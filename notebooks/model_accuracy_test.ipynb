```python
import os
import numpy as np
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing import image
from sklearn.metrics import accuracy_score, confusion_matrix

# 1. Load the trained CNN model from drive
model_path = "/content/drive/MyDrive/ds_task/cnn_product_model.h5"
cnn_model = load_model(model_path)

# 2. Define test dataset path and classes
test_dir = "/content/drive/MyDrive/ds_task/TEST_DATASETS"
classes = sorted(os.listdir(test_dir))  # assuming one folder per stock code

y_true, y_pred = [], []

# 3. Iterate over test images and predict
for class_name in classes:
    class_folder = os.path.join(test_dir, class_name)
    if not os.path.exists(class_folder):
        continue
    for img_file in os.listdir(class_folder):
        img_path = os.path.join(class_folder, img_file)
        img_array = image.img_to_array(image.load_img(img_path, target_size=(128,128))) / 255.0
        img_array = np.expand_dims(img_array, axis=0)

        pred_idx = np.argmax(cnn_model.predict(img_array, verbose=0))
        pred_class = classes[pred_idx]

        y_true.append(class_name)
        y_pred.append(pred_class)

# 4. Compute test accuracy
acc = accuracy_score(y_true, y_pred)
print("Test Accuracy:", acc)

# 5. Confusion matrix
cm = confusion_matrix(y_true, y_pred, labels=classes)
print("Confusion Matrix:\n", cm)
```
