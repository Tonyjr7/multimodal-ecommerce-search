```python
import json
import os
import requests
from tqdm import tqdm
from PIL import Image
from io import BytesIO

# Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')

# Path to dataset and scraped images in JSON files
json_folder = "/content/drive/MyDrive/ds_task/CNN_DATASETS"
output_folder = "/content/drive/MyDrive/ds_task/CNN_IMAGES"
os.makedirs(output_folder, exist_ok=True)

# Target size for images
target_size = (128, 128)  # width x height

for json_file in os.listdir(json_folder):
    if json_file.endswith(".json"):
        stock_code = os.path.splitext(json_file)[0]
        stock_folder = os.path.join(output_folder, stock_code)
        os.makedirs(stock_folder, exist_ok=True)

        with open(os.path.join(json_folder, json_file), 'r') as f:
            products = json.load(f)

        for i, product in enumerate(tqdm(products, desc=f"Downloading {stock_code}")):
            img_url = product.get("thumbnailImage")
            if img_url:
                try:
                    r = requests.get(img_url, timeout=10)
                    img = Image.open(BytesIO(r.content)).convert("RGB")
                    img = img.resize(target_size)
                    img_path = os.path.join(stock_folder, f"{i+1}.jpg")
                    img.save(img_path)
                except Exception as e:
                    print(f"Failed to download {img_url}: {e}")
```
